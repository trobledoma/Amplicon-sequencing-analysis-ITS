```python
Feature_counts = {}
Shannon = {}
Pielou = {}
Simpson = {}
Chao1 = {}

for column in seqOne.data:
    muestra = column[0]
    counts = seqOne.data[column]

    Feature_counts[muestra] = alpha.observed_otus(counts)
    Shannon[muestra] = alpha.shannon(counts)
    Pielou[muestra] = alpha.pielou_e(counts)
    Simpson[muestra] = alpha.simpson_e(counts)
    Chao1[muestra] = alpha.chao1(counts)

    print(f"Feature count in {muestra}: {Feature_counts[muestra]}")
    print(f"Shannon index in {muestra}: {Shannon[muestra]}")
    print(f"Pielou index in {muestra}: {Pielou[muestra]}")
    print(f"Simpson index in {muestra}: {Simpson[muestra]}")
    print(f"Chao1 index in {muestra}: {Chao1[muestra]}")
    print("-" * 40)
```

```python
import pandas as pd
import os

filepath = "/home/tatiana/PROYECTOS/fungil/analysis/"

# Leer datos
diversidad = pd.read_csv(
    os.path.join(filepath, "alphadiversity-fungi-table.csv")
)

# Índices de diversidad alfa
indices = [
    "Shannon index",
    "Feature_counts",
    "Pielou index",
    "Simpson index",
    "Chao1 index"
]

# Agrupar réplicas técnicas por muestra ambiental
grupos = diversidad.groupby(["location", "date"])

# Calcular estadísticos descriptivos
medias = grupos[indices].mean()
desviaciones = grupos[indices].std()
n_replicas = grupos[indices].count()
cv = (desviaciones / medias) * 100

# Construir una única tabla
resultados = pd.concat(
    [
        medias.add_suffix("_mean"),
        desviaciones.add_suffix("_SD"),
        n_replicas.add_suffix("_n"),
        cv.add_suffix("_CV_percent")
    ],
    axis=1
).reset_index()

# Redondear resultados numéricos
columnas_numericas = resultados.select_dtypes(include="number").columns
resultados[columnas_numericas] = resultados[columnas_numericas].round(3)

# Exportar CSV
archivo_salida = os.path.join(
    filepath, "variabilidad_tecnica_alfa_diversidad.csv"
)

resultados.to_csv(
    archivo_salida,
    index=False,
    encoding="utf-8-sig"
)

print("Archivo exportado correctamente:")
print(archivo_salida)
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Diversity indices
indices = [
    ("Feature_counts", "A) Feature counts"),
    #("Chao1 index", "B) Chao1 index"),
    ("Pielou index", "B) Pielou index"),
    ("Shannon index", "C) Shannon index"),
    ("Simpson index", "D) Simpson index")
]

# Create the figure: one row x four columns
fig = make_subplots(
    rows=1,
    cols=4,
    subplot_titles=[nombre for _, nombre in indices]
)

for col, (indice, nombre) in enumerate(indices, start=1):

    fig.add_trace(
        go.Box(
            x=diversidad["location"],
            y=diversidad[indice],
            name=nombre,
            boxpoints="all",  # Show all the points
            jitter=0.2,
            pointpos=0,
            showlegend=False
        ),
        row=1,
        col=col
    )

fig.update_layout(
    title="Figure 3",
    height=500,
    width=1500,
    template="plotly" # plotly, plotly_white, plotly_dark, seaborn, simple_white, presentation, xgridoff, ygridoof, gridon, none.
)

fig.show()
```
