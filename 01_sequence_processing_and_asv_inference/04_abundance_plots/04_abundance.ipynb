```python
# Define your file paths (update names if your files differ slightly)
counts_path = filepath + "feature_table.csv"          # Or your abundance/feature table
taxonomy_path = filepath + "taxonomy.csv"          # Your taxonomy table
metadata_path = filepath + "metadata.csv"
```

```python
# Load datasets
df_counts = pd.read_csv(counts_path, index_col=0)
df_tax = pd.read_csv(taxonomy_path, index_col=0)
metadata = pd.read_csv(metadata_path)
```

```python
# Check if sample names are in the index (rows) and transpose if needed
sample_id_col = "sample_name"
grouping_col = "location"
metadata[sample_id_col] = metadata[sample_id_col].astype(str).str.strip()

# If sample names are in the index, transpose the count table so rows = ASVs and columns = samples
if df_counts.index.isin(metadata[sample_id_col]).any():
    df_counts = df_counts.T
```

```python
# Assign Genus taxonomy to feature IDs (rows)
df_counts.index = df_tax.reindex(df_counts.index)["Genus"].fillna("Unclassified").values
```

```python
# Compute relative abundance (%) per sample
df_genus_counts = df_counts.groupby(level=0).sum()
df_relative = (df_genus_counts.div(df_genus_counts.sum(axis=0), axis=1)) * 100
```

```python
# Clean and map sample columns to metadata groups
df_relative.columns = df_relative.columns.astype(str).str.strip()
sample_to_group = dict(zip(metadata[sample_id_col], metadata[grouping_col]))
df_relative.columns = df_relative.columns.map(sample_to_group)
```

```python
# Average relative abundance across biological replicates per group
df_mean = df_relative.T.groupby(level=0).mean().T
```

```python
# Convert matrix to long format for Plotly
df_plot = df_mean.rename_axis("Genus").reset_index()
df_plot = df_plot.melt(
    id_vars="Genus",
    var_name="Location",
    value_name="Relative abundance (%)"
)
df_plot["Genus"] = df_plot["Genus"].str.replace(r"^g__", "", regex=True)

print(f"-> Success! Total rows in df_plot: {len(df_plot)}")
display(df_plot.head())
```

```python
# Treshold comparison groupoing low abundance groups
```

```python
thresholds = [0.1, 0.5, 1.0]
results = []

for umbral in thresholds:
    mean_abund = df_plot.groupby("Genus")["Relative abundance (%)"].mean()
    minor = mean_abund[mean_abund < umbral].index
    n_minor = len(minor)
    n_remaining = df_plot["Genus"].nunique() - n_minor + (1 if n_minor > 0 else 0)
    
    grouped_abund = (
        df_plot[df_plot["Genus"].isin(minor)]
        .groupby("Location")["Relative abundance (%)"]
        .sum()
    )
    
    row_data = {
        "Threshold (%)": umbral,
        "Genera grouped": n_minor,
        "Final categories": n_remaining
    }
    for loc, val in grouped_abund.items():
        row_data[f"{loc} (%)"] = val
        
    results.append(row_data)

comparison_table = pd.DataFrame(results).fillna(0)
print("=== THRESHOLD COMPARISON TABLE ===")
display(comparison_table.round(2))
```

```python
# Applying 0.5% as a filter for low abundance
```

```python
threshold = 0.5
mean_abundance = df_plot.groupby("Genus")["Relative abundance (%)"].mean()
minor_genera = mean_abundance[mean_abundance < threshold].index

df_plot_grouped = df_plot.copy()
df_plot_grouped["Genus"] = df_plot_grouped["Genus"].where(
    ~df_plot_grouped["Genus"].isin(minor_genera),
    f"Other genera (<{threshold}%)"
)

df_plot_grouped = (
    df_plot_grouped
    .groupby(["Location", "Genus"], as_index=False)
    ["Relative abundance (%)"]
    .sum()
)

print(f"-> Taxa grouped using {threshold}% threshold. Resulting categories: {df_plot_grouped['Genus'].nunique()}")
```

```python
# Plotting and figures
```

```python
import plotly.express as px

colors = [
    "#A6CEE3", "#E69F00", "#009E73", "#FB9A99", "#6A3D9A", 
    "#56B4E9", "#F0E442", "#8C564B", "#B2DF8A", "#D1495B", 
    "#D55E00", "#4D9221", "#F2B8D6", "#A23BEC", "#FF6F61", 
    "#BCBD22", "#BD1100", "#17BECF", "#7F8FE8", "#006D77", 
    "#FDBF6F", "#B19429"
]

group_minor = f"Other genera (<{threshold}%)"
group_unclassified = "Unclassified"

graphic_mean_abundance = df_plot_grouped.groupby("Genus")["Relative abundance (%)"].mean()

individual_genera = (
    graphic_mean_abundance
    .drop(index=[group_minor, group_unclassified], errors="ignore")
    .sort_values(ascending=False)
    .index.tolist()
)

genus_order = []
if group_unclassified in graphic_mean_abundance.index:
    genus_order.append(group_unclassified)
if group_minor in graphic_mean_abundance.index:
    genus_order.append(group_minor)
genus_order.extend(individual_genera)

color_map = {genus: colors[i % len(colors)] for i, genus in enumerate(individual_genera)}

fig = px.bar(
    df_plot_grouped,
    x="Location",
    y="Relative abundance (%)",
    color="Genus",
    barmode="stack",
    color_discrete_map=color_map,
    category_orders={"Genus": genus_order},
    title="Mean Relative Abundance of Fungal Genera by Location",
    labels={
        "Location": "Location",
        "Relative abundance (%)": "Mean relative abundance (%)",
        "Genus": "Fungal genus"
    }
)

fig.update_traces(marker_line_width=0)

fig.for_each_trace(
    lambda trace: trace.update(
        name=f"<i>{trace.name}</i>"
        if trace.name not in [group_minor, group_unclassified]
        else trace.name
    )
)

fig.update_layout(
    template="plotly_white",
    width=1200,
    height=900,
    font=dict(size=16),
#    xaxis_title="Location",
    yaxis_title="Mean relative abundance (%)",
    yaxis=dict(range=[0, 100]),
    legend_title_text="",
    legend=dict(
        orientation="v",
        x=1.02,
        xanchor="left",
        y=1,
        yanchor="top",
        font=dict(size=14),
        itemsizing="constant",
        traceorder="reversed"
    ),
    margin=dict(l=80, r=300, t=60, b=80)
)

# Display interactive chart and export files
fig.show()

fig.write_image(filepath + "fungal_genera_abundance.pdf")
fig.write_image(filepath + "funga_genera_abundance.png", scale=3)
```
